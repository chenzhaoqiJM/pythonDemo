```python
# myhsv
itmax = np.max(img, axis=2)
itmin = np.min(img, axis=2)
V = itmax
S = (itmax - itmin)/itmax
```

```python
plt.figure(figsize=(9,12))
plt.subplot(1,2,1)
plt.imshow(S, cmap='gray')
plt.subplot(1,2,2)
plt.imshow(V, cmap='gray')
```

```python
B,G,R = img[:,:,0]/255, img[:,:,0]/255, img[:,:,0]/255
```

```python
H = np.ones(R.shape, dtype=np.uint8)
diffit = itmax-itmin
_mask0 = diffit !=0

mask = R==itmax
mask = mask&_mask0
H[mask] = ((G-B)[mask]/(diffit)[mask]*60)

mask = G==itmax
mask = mask&_mask0
H[mask] = (120+(B-R)[mask]/(diffit)[mask]* 60)

mask = B==itmax
mask = mask&_mask0
H[mask] = (240 +(R-G)[mask]/(diffit)[mask]* 60)

_mask = H<0
H[_mask] = H[_mask]+360
```

```python
plt.figure(figsize=(9,12))
plt.imshow(H, cmap='gray')
```
