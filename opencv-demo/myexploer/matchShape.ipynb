```python
import numpy as np
import cv2
import time
import matplotlib.pyplot as plt
```

```python
path1 = "C:\\Users\\86994\\Desktop\\1.png"
path2 = "C:\\Users\\86994\\Desktop\\2.png"

image1 = cv2.imread(path1)
image2 = cv2.imread(path2)

print(image1.shape)
print(image2.shape)
```

```python
img1 = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY) #灰度化
img2 = cv2.cvtColor(image2, cv2.COLOR_BGR2GRAY) #灰度化
```

```python
th = 150
retval, img_binary1 = cv2.threshold(img1, th, 255, cv2.THRESH_BINARY) #
retval, img_binary2 = cv2.threshold(img2, th, 255, cv2.THRESH_BINARY) #
```

```python
img1_edge = cv2.Laplacian(img_binary1, -1)
img2_edge = cv2.Laplacian(img_binary2, -1)
plt.imshow(img1_edge, cmap='gray')
# plt.imshow(img2_edge, cmap='gray')
plt.show()
```

```python
contours1, hierarchy1	= cv2.findContours(img1_edge, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
contours2, hierarchy2	= cv2.findContours(img2_edge, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
```

```python
src_img = image1.copy()
cv2.drawContours(src_img, contours1, 0, (255,0,0), thickness=5) #画原轮廓
plt.imshow(src_img)
plt.show()
```

```python
src_img = image2.copy()
cv2.drawContours(src_img, contours2, 0, (255,0,0), thickness=5) #画原轮廓
plt.imshow(src_img)
plt.show()
```

```python
print("轮廓周长：", cv2.arcLength(contours1[0],closed=True), " -- ", cv2.arcLength(contours2[0],closed=True))
```

```python
matchRes = cv2.matchShapes(contours1[0], contours2[0], method=1, parameter=0)
print(matchRes)
```
