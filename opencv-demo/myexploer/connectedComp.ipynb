```python
import numpy as np
import cv2
import time
import matplotlib.pyplot as plt
```

```python
path1 = 'C:\\Users\\hp\\Desktop\\Image_20230227105332742.bmp'
image = cv2.imread(path1)
print(image.shape)
```

```python
img1 = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) #灰度化
plt.imshow(img1, cmap='gray')
plt.show()
```

```python
th = 100
retval, img_binary = cv2.threshold(img1, th, 255, cv2.THRESH_BINARY) #
plt.imshow(img_binary, cmap='gray')
plt.show()
```

```python
retval, labels = cv2.connectedComponents(img_binary)
```

```python
drawImg = np.ones_like(img_binary, dtype=np.uint8)*255
drawImg[labels==1] = 0
plt.imshow(drawImg,cmap='gray')
```
