```python
# License: BSD
# Author: Sasank Chilamkurthy

from __future__ import print_function, division

import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim import lr_scheduler
import numpy as np
import torchvision
from torchvision import datasets, models, transforms
import matplotlib.pyplot as plt
import time
import os
import copy
from PIL import Image
plt.ion()   # interactive mode
```

```python
transform = transforms.Compose([
        transforms.Resize(224),
        # transforms.CenterCrop(224),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
```

```python
img=Image.open('E:/datasets/tire_dataset/leb/leb_1.jpg')
tensorImg = transform(img)
batch = tensorImg
batch = batch.reshape([1, batch.shape[0], batch.shape[1], batch.shape[2]]) 
print(batch.shape)
```

```python
from my_model_class import myAlexNet, myMobileNetV2, myHelper
```

```python
model_ft = myAlexNet()
model_ft.load_state_dict(torch.load('C:\\Users\\hp\\Desktop\\model_alexnet_static.pth'))
model_ft.eval()
output = model_ft(batch)
print(output)
```

```python
traced = torch.jit.trace(model_ft, torch.rand(1, 3, 224, 224) )
traced.save('C:\\Users\\hp\\Desktop\\wrapped_alexnet.pt')
loaded = torch.jit.load('C:\\Users\\hp\\Desktop\\wrapped_alexnet.pt')
print(loaded(batch))
```
