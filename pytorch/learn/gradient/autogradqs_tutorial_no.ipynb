```python
import torch

x = torch.ones(5)  # input tensor
y = torch.zeros(3, requires_grad=True)  # expected output
w = torch.randn(5, 3, requires_grad=True)
b = torch.randn(3, requires_grad=True)
z = torch.matmul(x, w)+b
loss = torch.nn.functional.binary_cross_entropy_with_logits(z, y)
```

```python
print(f"Gradient function for z = {z.grad_fn}")
print(f"Gradient function for loss = {loss.grad_fn}")
```

```python
loss.backward()
print(w.grad)
print(b.grad)
```

```python
print(y.grad)
```

```python
c = torch.tensor([1,2,3,4],requires_grad=True, dtype=torch.float32)
f = torch.tensor([[4,5,7,1], [1,3,2,1]],requires_grad=True, dtype=torch.float32)
f = f.T
k = torch.matmul(c, f)

s = torch.tensor([1,1], dtype=torch.float32)
s = s.T

res = torch.matmul(k, s)
res.backward()

print(f.grad)
print(c.grad)
```
