```python
# 初始化向量

# v0 = np.array([[-1,1,-1,1, -1,1,1,1]], np.float32)
v0 = np.array([[1,-1,1,-1]], np.float32)
v0 = v0.T
```

```python
def Energy(W, v, theta):
    _e = -0.5 * (v.T @ (W @ v)) + v.T @ theta
    return _e

def state(W, v, theta):
    return W@v- theta
```

```python
# 第一次迭代
_1E = Energy(W, v0, Theta)
_1u = state(W, v0, Theta)
```
