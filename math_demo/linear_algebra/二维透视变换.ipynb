```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = 10
b = 0.
```

```python
# 透视变换矩阵，行向量形式
M = [
    [1., 0., a],
    [0., 1., b],
    [0., 0., 1.]]
M = np.array(M, dtype=np.float32)
```

```python
AB = np.linspace(0,1,50)
AB = np.array([[0, a, 1] for a in AB], dtype=np.float32)
```

```python
BC = np.linspace(0,1,50)
BC = np.array([[_a, 1, 1] for _a in BC], dtype=np.float32)
```

```python
CD = np.linspace(0,1,50)
CD = np.array([[1, _a, 1] for _a in CD], dtype=np.float32)
```

```python
DA = np.linspace(0,1,50)
DA = np.array([[_a, 0, 1] for _a in DA], dtype=np.float32)
```

```python
plt.figure(figsize=(5,5))
plt.scatter(AB[:,0], AB[:,1])
plt.scatter(BC[:,0], BC[:,1])
plt.scatter(CD[:,0], CD[:,1])
plt.scatter(DA[:,0], DA[:,1])
```

```python
AB1, BC1, CD1, DA1 = AB@M, BC@M, CD@M, DA@M
```

```python
for _item in [AB1, BC1, CD1, DA1]:
    for _subi in range(0, len(_item)):
        _item[_subi] = _item[_subi]/ _item[_subi][2]
```

```python
fig = plt.figure(figsize=(5,5))
ax1=fig.add_subplot(1,1,1) 
plt.xlim((0.0, 1.0))
plt.ylim((0, 1.0))
for _a in [0.,0.1, 0.2, 0.3, 0.4,0.6,0.8,0.9,0.95, 1.,2.,3.,4., 5,10, 20]:
    _M = [
    [1., 0., _a],
    [0., 1., 0],
    [0., 0., 1.]]
    _M = np.array(_M, dtype=np.float32)
    _AB1, _BC1, _CD1, _DA1 = AB@_M, BC@_M, CD@_M, DA@_M
    for _item in [_AB1, _BC1, _CD1, _DA1]:
        for _subi in range(0, len(_item)):
            _item[_subi] = _item[_subi]/ _item[_subi][2]
  
    ax1.scatter(_AB1[:,0], _AB1[:,1])
    ax1.scatter(_BC1[:,0], _BC1[:,1])
    ax1.scatter(_CD1[:,0], _CD1[:,1])
    ax1.scatter(_DA1[:,0], _DA1[:,1])
plt.show()
```
