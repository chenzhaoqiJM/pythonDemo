```python
# -*- coding: utf-8 -*-
"""
Created on Thu Jul 15 10:16:35 2021

@author: zq
"""

import pandas as pd
import matplotlib.pyplot as plt
import xlrd
import xlwt
import os
import openpyxl

def read_excel(file_path, path):
    # 打开文件
    workBook = xlrd.open_workbook(file_path);
    ## 2.1 法1：按索引号获取sheet内容
    sheet1_content1 = workBook.sheet_by_index(0); # sheet索引从0开始

    colsx = sheet1_content1.col_values(0)# 获取第1列内容
    colsy = sheet1_content1.col_values(1)
    colsz = sheet1_content1.col_values(2)

    i1 = sheet1_content1.col_values(3)
    i2 = sheet1_content1.col_values(4)
    i3 = sheet1_content1.col_values(5)
    
    #容量更大
    f = openpyxl.Workbook()
    sheet1 = f.create_sheet(index=0)
    for i in range(1,len(colsx)):
        sheet1.cell(column=1, row=i, value = i1[i])
        sheet1.cell(column=2, row=i, value = i2[i])
        sheet1.cell(column=3, row=i, value = i3[i])
        
        sheet1.cell(column=4, row=i, value = colsx[i]* 100000.0 + 500.0)
        sheet1.cell(column=5, row=i, value = colsy[i]* 100000.0 + 500.0)
        sheet1.cell(column=6, row=i, value = colsz[i]* 100000.0 + 500.0)


    base_path = "C:\\Users\\hp\\Desktop\\"    #保存的路径
    save_path = os.path.join(base_path, path)   #拼接成完整路径
    f.save(save_path)
    
    return ' ' 



path1 = '三维力.xlsx'
path2 = '三维力改.xlsx'
read_excel(path1, path2)
```
