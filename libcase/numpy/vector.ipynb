```python
import numpy as np
import time
```

```python
# points = np.ones([2000, 2], dtype=np.float32)
points = np.random.random([28000,3])
```

```python
points_filter = []
points_copy = points.copy()
```

```python
t1 = time.time()
while len(points_copy)!=0:
    points_filter.append(points_copy[0])  # 中心点放入列表
    dis_vec = points_copy - points_copy[0]
    dis_vec_pow = np.power(dis_vec, 2)
    dis_vec_pow_sum = np.sum(dis_vec_pow, axis=1)
    points_copy = points_copy[dis_vec_pow_sum>=0.4] # 半径平方大于0.4的点保留
    print("count -----------------")
print(time.time()-t1)
```

```python
points_filter = np.array(points_filter)
print(points_filter)
```
