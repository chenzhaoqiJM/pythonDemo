```python
from scipy import integrate
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 被积分函数
def myfunction(x):
    # you need to define result
    return x

# 积分函数、积分下限、积分上限
integral = integrate.quad(myfunction, 0.0, 4.0)
print(integral)
```

```python
_x2 = lambda x: x**2
_res = integrate.quad(_x2, 0, 4)
print(_res)
print(4**3 / 3.)  # analytical result(解析解)
```

```python
invexp = lambda x: np.exp(-x)
_res = integrate.quad(invexp, 0, np.inf) # 0 到正无穷积分
print(_res)
```

```python
# 带系数的积分
f = lambda x,a : a*x
y, err = integrate.quad(f, 0, 1, args=(1,)) # a=1时
print(y)
y, err = integrate.quad(f, 0, 1, args=(3,)) # a=3时
print(y)
```

请注意，与积分间隔的大小相比，脉冲形状和其他尖锐特征可能无法使用这种方法正确地积分。这个限制的一个简化的例子是对y轴反射阶跃函数积分，在积分范围内有许多0值。

```python
# 分段函数,积分范围很多0值
f = lambda x: 1 if x<=0 else 0 # 小于等于0为1，大于0为0
y, err =integrate.quad(f, -1, 1)
print(y)
y, err = integrate.quad(f, -1, 100)
print(y)
y, err = integrate.quad(f, -1, 1000)
print(y, err)
```

向量值函数自适应积分

```python
# 批量化计算
alpha = np.linspace(0.0, 2.0, num=30)
f = lambda x: x**alpha
y, err = integrate.quad_vec(f, 0, 2)
plt.plot(alpha, y)
plt.xlabel(r"$\alpha$")
plt.ylabel(r"$\int_{0}^{2} x^\alpha dx$")
plt.show()
```
