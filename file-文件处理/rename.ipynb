```python
import cv2
import numpy as np
import os
import sys
import shutil
```

```python
source_imgs_dir = 'C:\\Users\\86994\\Desktop\\kg\\'
source_imgs_name_lists = os.listdir(source_imgs_dir)
```

```python
new_imgs_save_dir = 'C:\\Users\\86994\\Desktop\\500x500kg\\'
```

```python

for i, img_name in enumerate(source_imgs_name_lists):
    path = os.path.join(source_imgs_dir, img_name)
    img = cv2.imread(path)
    img = cv2.resize(img, [500,500])
    save_path = os.path.join(new_imgs_save_dir, str(i)+'.jpg')
    cv2.imwrite(save_path, img)
```
