```python
import cv2
import numpy as np
import os
import sys
import shutil
```

```python
source_imgs_dir = 'C:\\Users\\hp\\Desktop\\4\\test'
source_imgs_name_lists = os.listdir(source_imgs_dir)
```

```python
imgs_name = []
for item in source_imgs_name_lists:
    if item.split('.')[-1] in ['jpg', 'png', 'bmp', 'jpeg']:
        imgs_name.append(item)
```

```python
new_imgs_save_dir = 'C:\\Users\\hp\\Desktop\\4\\test\\rect'
```

```python
# 定大小放缩
ratio = 3
for i, img_name in enumerate(imgs_name):
    path = os.path.join(source_imgs_dir, img_name)
    img = cv2.imread(path)
    img = cv2.resize(img, [1200,900])
    # img = cv2.resize(img, [900,1200])
    save_name = img_name
    # save_name = str(i)+'.jpg'
    save_path = os.path.join(new_imgs_save_dir, save_name)
    cv2.imwrite(save_path, img)
```

```python
# 定大小裁剪
for i, img_name in enumerate(imgs_name):
    path = os.path.join(source_imgs_dir, img_name)
    img = cv2.imread(path)
    img_rect = img[323:1634, 969:2330, :]
    save_path = os.path.join(new_imgs_save_dir, img_name.split('.')[0]+'.jpg')
    cv2.imwrite(save_path, img_rect)
```
