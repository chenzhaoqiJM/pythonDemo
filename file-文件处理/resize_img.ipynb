```python
import cv2
import numpy as np
import os
import sys
import shutil
```

```python
source_imgs_dir = 'C:\\Users\\hp\\Desktop\\imgs\\'
source_imgs_name_lists = os.listdir(source_imgs_dir)
```

```python
imgs_name = []
for item in source_imgs_name_lists:
    if item.split('.')[-1] in ['jpg', 'png', 'bmp', 'jpeg']:
        imgs_name.append(item)
```

```python
for i in range(1, len(source_imgs_name_lists)+1):
    try:
        index = source_imgs_name_lists[i-1].find('.') #找到后缀名所在之索引
        name_suffix = source_imgs_name_lists[i-1][index:] #提取后缀名
        # print(name_suffix)
        new_file_name = str(i) + name_suffix #修改为数字加后缀名
        
        ###文件重命名，使用path.join保证路径拼接正确
        os.rename(os.path.join(source_imgs_dir , source_imgs_name_lists[i-1]), os.path.join(source_imgs_dir , new_file_name) )
        
    except :
        print("Please check the format !!!") #发生异常打印信息
```

```python
new_imgs_save_dir = 'C:\\Users\\hp\\Desktop\\imgs_b2\\'
```

```python
# 定大小放缩
ratio = 3
for i, img_name in enumerate(imgs_name):
    path = os.path.join(source_imgs_dir, img_name)
    img = cv2.imread(path)
    img = cv2.resize(img, [1200,900])
    # img = cv2.resize(img, [900,1200])
    save_name = img_name
    # save_name = str(i)+'.jpg'
    save_path = os.path.join(new_imgs_save_dir, save_name)
    cv2.imwrite(save_path, img)
```
