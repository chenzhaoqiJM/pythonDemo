```python
import numpy as np
import cv2
import time
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
mean_iou = 95.0 # 平均交并比的百分比
```

```python
edge_line = np.math.sqrt(mean_iou)
x = np.math.sqrt(100-mean_iou+edge_line**2)-edge_line
```

```python
img = np.ones(shape=(800, 800, 4), dtype=np.uint8)*255
draw_box_len = 480
start_point1 = ((img.shape[0]-draw_box_len)//2, (img.shape[1]-draw_box_len)//2)
end_point1 = (start_point1[0]+480, start_point1[1]+480)
_ = cv2.rectangle(img,start_point1, end_point1, (122,0,134, 80),-1)
```

```python
start_point2 = (start_point1[0]+ int(draw_box_len*x/(x+edge_line)), start_point1[1]+ int(draw_box_len*x/(x+edge_line)))
end_point2 = (start_point2[0]+480, start_point2[1]+480)
_ = cv2.rectangle(img,start_point2, end_point2, (12,134,14, 30),-1)
```
