```python
import numpy as np
import cv2
import time
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
src_mask_path = "C:\\Users\\hp\\Desktop\\4_rect_1\\1024\\gtFine"
```

```python
imgs_name = []
for _item in os.listdir(src_mask_path):
    if _item.split('.')[-1] in ['jpg', 'png', 'bmp', 'jpeg']:
        imgs_name.append(_item)
```

```python
object_sum = 0.
total_sum = 0.
for name in imgs_name:
    mask = Image.open(os.path.join(src_mask_path, name))
    mask = np.array(mask)
    total_sum = total_sum + mask.shape[0]*mask.shape[1]
    object_sum = object_sum + np.count_nonzero(mask == 1)
```

```python
ratio = object_sum / total_sum
```
