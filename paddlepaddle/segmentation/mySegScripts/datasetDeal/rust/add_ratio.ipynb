```python
import numpy as np
import cv2
import time
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
img = np.array(Image.open(r'C:\Users\hp\Desktop\tc_new\2\resized\71.jpg'))
roi = np.array(Image.open(r'C:\Users\hp\Desktop\tc_new\2\71.png'))
rust = np.array(Image.open(r'C:\Users\hp\Desktop\tc_new\2\71_rust.png'))
```

```python
roi_mask = roi==1
rust_mask = rust==1
_img_copy1 = img.copy()
_img_copy1 = _img_copy1.astype(np.float32)/255.
_img_copy1[roi_mask] = _img_copy1[roi_mask] + (0,0.6,0)
_add_mask = np.logical_and(roi_mask, rust_mask)
_img_copy1[_add_mask] = _img_copy1[_add_mask] + (0.7,0,0)

print('ratio: ', float(np.count_nonzero(_add_mask))/np.count_nonzero(roi_mask))
# plt.figure(figsize=(9,18))
# plt.subplot(1,2,1)
# plt.imshow(img_gray, cmap='gray')
# plt.subplot(1,2,2)
# plt.imshow(img_binary, cmap='gray')
plt.imshow(_img_copy1)
```
