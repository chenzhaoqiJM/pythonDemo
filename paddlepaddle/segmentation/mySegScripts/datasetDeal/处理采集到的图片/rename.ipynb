```python
import cv2
import numpy as np
import os
import sys
import shutil
```

```python
source_imgs_dir =  r'C:\Users\hp\Desktop\tc_data\forRust\imgs_5\crop'
start_num = 138
```

```python
def reName(source_imgs_dir):
    # 映射到非数字名称
    source_imgs_name_lists = os.listdir(source_imgs_dir)
    imgs_name = []
    for item in source_imgs_name_lists:
        if item.split('.')[-1] in ['jpg', 'png', 'bmp', 'jpeg']:
            imgs_name.append(item)
    for i in range(0, len(source_imgs_name_lists)):
        try:
            name_suffix = source_imgs_name_lists[i].split('.')[-1] #提取后缀名
            # print(name_suffix)
            new_file_name = str(i)+"aadsaq" +"."+ name_suffix #修改为数字加后缀名
            
            ###文件重命名，使用path.join保证路径拼接正确
            os.rename(os.path.join(source_imgs_dir , source_imgs_name_lists[i]), os.path.join(source_imgs_dir , new_file_name) )
            
        except Exception as e:
            print(e,'\n')
            # print("Please check the format !!!") #发生异常打印信息
    
    # 到数字名称
    source_imgs_name_lists = os.listdir(source_imgs_dir)
    imgs_name = []
    for item in source_imgs_name_lists:
        if item.split('.')[-1] in ['jpg', 'png', 'bmp', 'jpeg']:
            imgs_name.append(item)
    for i in range(0, len(source_imgs_name_lists)):
        try:
            name_suffix = source_imgs_name_lists[i].split('.')[-1] #提取后缀名
            # print(name_suffix)
            new_file_name = str(i+start_num) +"."+ name_suffix #修改为数字加后缀名
            
            ###文件重命名，使用path.join保证路径拼接正确
            os.rename(os.path.join(source_imgs_dir , source_imgs_name_lists[i]), os.path.join(source_imgs_dir , new_file_name) )
            
        except Exception as e:
            print(e,'\n')
            # print("Please check the format !!!") #发生异常打印信息
```

```python
reName(source_imgs_dir)
```
