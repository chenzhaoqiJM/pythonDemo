```python
import numpy as np
import cv2
import time
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
def getRect(path):
    roi_mask = np.array(Image.open(path))
    binary = roi_mask*255
    points = cv2.findNonZero(binary)
    rect = cv2.boundingRect(points)
    return(rect)
```

```python
path = r'C:\Users\hp\Desktop\tc_data\150x70\blister4x\new1\78\src\annotations\78_1.png'
rect = getRect(path)
```

```python
source_imgs_dir = r'C:\Users\hp\Desktop\tc_data\150x70\blister4x\new1\78\src'
imgs_name = []
for item in os.listdir(source_imgs_dir):
    if item.split('.')[-1] in ['jpg', 'png', 'bmp', 'jpeg']:
        imgs_name.append(item)
```

```python
new_imgs_save_dir = r'C:\Users\hp\Desktop\tc_data\150x70\blister4x\new1\78\rect'
```

```python
for i, img_name in enumerate(imgs_name):
    path = os.path.join(source_imgs_dir, img_name)
    img = cv2.imread(path)
    
    sub_img = img[rect[1]:rect[1]+rect[3], rect[0]:rect[0]+rect[2] ,:]
    save_name = img_name.split('.')[0]+'.jpg'
    save_path = os.path.join(new_imgs_save_dir, save_name)
    cv2.imwrite(save_path, sub_img)
```
